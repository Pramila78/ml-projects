```python
import pandas as pd
import numpy as np
train_d=pd.read_csv("train.csv")
train_d.head()
```

```python
test_d=pd.read_csv("test.csv")
test_d.head()
```

```python
women=train_d.loc[train_d.Sex=='female']["Survived"]
rate_women=sum(women)/len(women)
rate_women
```

```python
men=train_d.loc[train_d.Sex=='male']["Survived"]
rate_men=sum(men)/len(men)
rate_men
```

```python
from sklearn.ensemble import RandomForestClassifier
y=train_d["Survived"]
features=["Pclass", "Sex", "SibSp", "Parch"]
x=pd.get_dummies(train_d[features])
x_test=pd.get_dummies(test_d[features])
m=RandomForestClassifier(n_estimators=100,max_depth=5,random_state=1)
m.fit(x,y)
predictions=m.predict(x_test)
```

```python
predictions
```

```python
output=pd.DataFrame({'PassengerId':test_d.PassengerId,'Survived':predictions})
output
```

```python
output.to_csv('submission.csv',index=False)
print("Your submission was successfully saved!")
```
